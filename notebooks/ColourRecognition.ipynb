## _Renk CSV'sini oku_

```python
import cv2
import pandas as pd
import numpy as np
import os
```

```python
# CSV sütunları: color, color_name, hex, R, G, B
index = ["color", "color_name", "hex", "R", "G", "B"]
csv = pd.read_csv('../data/colors.csv', names=index, header=None)
```

## _Test resmini yükle_

```python
img = cv2.imread("../data/colour.jpeg")
```

## _Orijinal resmi sakla_

```python
img_copy = img.copy()
```

## _Global değişkenler_

```python
r = g = b = xpos = ypos = 0
display_text = ""  # ekranda sabit kalacak yazı
```

## _Renk tanıma fonksiyonu_

```python
def recognize_color(R, G, B):
    minimum = 10000
    cname = ""
    for i in range(len(csv)):
        d = abs(R-int(csv.loc[i,"R"])) + abs(G-int(csv.loc[i,"G"])) + abs(B-int(csv.loc[i,"B"]))
        if d <= minimum:
            minimum = d
            cname = csv.loc[i,"color_name"]
    return cname
```

## _Fare tıklama fonksiyonu (tek tık)_

```python
def mouse_click(event, x, y, flags, param):
    global b, g, r, xpos, ypos, display_text
    if event == cv2.EVENT_LBUTTONDOWN:
        xpos, ypos = x, y
        b, g, r = img[y, x]
        b, g, r = int(b), int(g), int(r)
        # Tıklayınca display_text güncellenir
        display_text = recognize_color(r, g, b) + f'  R={r} G={g} B={b}'
```

## _OpenCV pencere_

```python
cv2.namedWindow('Color Recognition App')
cv2.setMouseCallback('Color Recognition App', mouse_click)

while True:
    # Her frame orijinal resmin kopyası
    img_temp = img_copy.copy()

    # Eğer display_text doluysa yazıyı göster
    if display_text:
        cv2.rectangle(img_temp, (20,20), (750,60), (b,g,r), -1)
        color = (0,0,0) if r+g+b > 600 else (255,255,255)
        cv2.putText(img_temp, display_text, (50,50), cv2.FONT_HERSHEY_SIMPLEX, 0.8, color, 2, cv2.LINE_AA)

    cv2.imshow('Color Recognition App', img_temp)

    # ESC tuşu ile çık
    if cv2.waitKey(20) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```
